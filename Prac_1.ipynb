```python
df['Species'] = df['Species'].astype('category')
```

```python
columns = df.columns
columns
df.dtypes
```

```python
for i in columns:
    if i != 'Species':
        df[i] = df[i].fillna(df[i].mean())
        
```

```python
df['Species'] = df['Species'].fillna(df['Species'].mode()[0])
```

```python
df.shape
```

```python
df['SepalLengthCm'] = (df['SepalLengthCm']-df['SepalLengthCm'].min())/(df['SepalLengthCm'].max()-df['SepalLengthCm'].min())
```
