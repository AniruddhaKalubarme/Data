```python
import pandas as pd
import numpy as np
```

```python
Data = {
    'Age' : ['0-10', '11-20', '21-30', '31-40', '41-50', '0-10', '11-20', '21-30', '31-40', '41-50', '0-10', '11-20', '21-30', '31-40', '41-50'],
    'Salary' : [0,10000,200000,40000,50000,2500,12000,20000,4000,5000,500,1000,20000,4000,5000]
}
```

```python
df = pd.DataFrame(Data)
```

```python
df
```

```python
group = df.groupby('Age')['Salary']
```

```python
group
```

```python
summery = group.agg(['min', 'max', 'mean', 'std', 'median'])
```

```python
print(summery)
```
